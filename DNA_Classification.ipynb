```python
numeric_df=pd.get_dummies(Df).astype(int)
numeric_df.iloc[:5]
```

# spliting dataset:

```python
from sklearn.model_selection  import train_test_split
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier,AdaBoostClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import confusion_matrix,classification_report,accuracy_score
```

```python
x=df.drop('class',axis=1)
y=df['class']
```

```python
numeric_df.columns
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
names=['knn','dt','rf','ada','svm_linear','svm_rbf','svm_sigmoid','naive']
clasifiers=[KNeighborsClassifier(n_neighbors=3),DecisionTreeClassifier(max_depth=5),RandomForestClassifier(max_depth=5,n_estimators=10),
            AdaBoostClassifier(),SVC(kernel='linear'),SVC(kernel='rbf'),SVC(kernel='sigmoid'),GaussianNB()]
```

```python
from sklearn.model_selection import KFold,cross_val_score
```

```python
models=list(zip(names,clasifiers))
print(models)
```

```python
names=[]
results=[]
val=[]
for name ,model in models:
    # print(model)
    kfold=KFold(n_splits=10)
    cv_results=cross_val_score(model,xtrain,ytrain,cv=kfold,scoring='accuracy')
    results.append(cv_results)
    names.append(name)
    mean_score=round(cv_results.mean(),2)
    value=f'{name} :  {round(cv_results.mean(),2)} [{cv_results.std()}]'
    print(value,"\n")
    val.append(mean_score)
```

```python
score=pd.DataFrame(val,index=names,columns=['cv_results'])
```

```python
score
```

```python
new_acc=[]
for noun,verb in models:
    fits=verb.fit(xtrain,ytrain)
    ypred=fits.predict(xtest)
    cm=confusion_matrix(ytest,ypred)
    accss=round(accuracy_score(ytest,ypred),2)
    clf=classification_report(ytest,ypred)
    print(f"{noun} {accss} : \n{cm}\n\t{clf}\n\n")
    new_acc.append(accss)
```

```python
score['test_accuracy']=new_acc
```
